# **Imputación de Datos Multivariantes**

```python
!pip install miceforest
```

```python
#Imputación de datos = Rellenado de celdas Null con Valores por defecto
#con Datos Multivariantes
#se instala la libreria miceforest (para aplicarle el algoritmo miceforest)
#'!pip install miceforest'
#Se importa como
import miceforest as mf
from  sklearn.datasets import load_iris #importamos load_iris
import numpy as np
```

```python
load_iris
```

```python
from sklearn import datasets
import pandas as pd
from sklearn import datasets #Atraigo el data set iris del la biblioteca Sklearn
import pandas as pd
import numpy as np
iris = datasets.load_iris() #Cargo el iris

iris = pd.DataFrame(data =np.c_[iris["data"], iris["target"]],
                    columns = iris['feature_names'] + ['target']) #Convierto el formato de database de iris en Dataset de pandas e añado la columna target
iris.rename({"target" : 'species'}, axis = 1, inplace = True) # Cambio el nombre de la columna target por  Specie
iris
```

```python
#renombramos la columna target para species

iris['species'] = iris['species'].astype('category') #Cambio el tipo de la columna iris por la variable categoria
iris.info()
```

```python
## Asigno valores nuelos
iris_amp  =mf.ampute_data(iris, perc = 0.1, random_state = 50) ##realizo una amputación de valores aleatorios se le dice que el perc = 0.1 (el 10% de los datos se quitaran) y el random_state es la profundidad del arbol de decisión
iris_amp
```

```python
#Crear Kernel.
kds = mf.ImputationKernel(
    iris_amp,       #Nombre del dataframe para entrenar el modelo
    random_state=50   ##Valores aleatorios para entrenar el modelo
)


#Correr el algoritmo MICE durante 2 iteracciones
kds.mice(2)

##Retorna el data set completo
iris_total = kds.complete_data()
iris_total
```
